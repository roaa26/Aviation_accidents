# Aviation accidents project

In this project I clean the aviation accidents data from Kaggle. I will do it step by step with pandas.

First I import pandas. It is the only library I need.

```python
import pandas as pd
```

Now I read the csv file and I look at the first 5 rows.

```python
df = pd.read_csv('airline_accidents.csv', low_memory=False)
df.head()
```

I want to know how many rows and how many columns are in the file.

```python
df.shape
```

Now I check the type of every column.

```python
df.info()
```

Now I count the missing values.

```python
df.isnull().sum()
```

I looked at the data and I saw that many columns are almost empty.
like Air Carrier and Schedule.
They are not useful for my analysis.
so I will drop them.
The last column Unnamed: 30 is completely empty so I will drop it too.

```python
df = df.drop(columns=['Latitude', 'Longitude', 'Airport Code', 'Airport Name', 'Aircraft Category', 'Registration Number', 'Schedule', 'Air Carrier', 'FAR Description', 'Purpose of Flight', 'Report Publication Date', 'Unnamed: 30'])
```

The file has spaces around every word like ' VMC ' and ' 1 '. I will remove these spaces from the text columns.

```python
df['Event Id'] = df['Event Id'].str.strip()
df['Accident Number'] = df['Accident Number'].str.strip()
df['Investigation Type'] = df['Investigation Type'].str.strip()
df['Event Date'] = df['Event Date'].str.strip()
df['Location'] = df['Location'].str.strip()
df['Country'] = df['Country'].str.strip()
df['Injury Severity'] = df['Injury Severity'].str.strip()
df['Aircraft Damage'] = df['Aircraft Damage'].str.strip()
df['Make'] = df['Make'].str.strip()
df['Model'] = df['Model'].str.strip()
df['Amateur Built'] = df['Amateur Built'].str.strip()
df['Number of Engines'] = df['Number of Engines'].str.strip()
df['Engine Type'] = df['Engine Type'].str.strip()
df['Weather Condition'] = df['Weather Condition'].str.strip()
df['Broad Phase of Flight'] = df['Broad Phase of Flight'].str.strip()
```

Some places are still empty after the spaces are gone. I will put the word Unknown in these places.

```python
df['Investigation Type'] = df['Investigation Type'].replace('', 'Unknown')
df['Location'] = df['Location'].replace('', 'Unknown')
df['Country'] = df['Country'].replace('', 'Unknown')
df['Aircraft Damage'] = df['Aircraft Damage'].replace('', 'Unknown')
df['Make'] = df['Make'].replace('', 'Unknown')
df['Model'] = df['Model'].replace('', 'Unknown')
df['Amateur Built'] = df['Amateur Built'].replace('', 'Unknown')
df['Number of Engines'] = df['Number of Engines'].replace('', 'Unknown')
df['Engine Type'] = df['Engine Type'].replace('', 'Unknown')
df['Weather Condition'] = df['Weather Condition'].replace('', 'Unknown')
df['Broad Phase of Flight'] = df['Broad Phase of Flight'].replace('', 'Unknown')
```

The injury columns are numbers but the file keeps them as text like ' 1 '. I will change them to real numbers. The empty places will be 0 because the report has no number.

```python
df['Total Fatal Injuries'] = pd.to_numeric(df['Total Fatal Injuries'], errors='coerce').fillna(0)
df['Total Serious Injuries'] = pd.to_numeric(df['Total Serious Injuries'], errors='coerce').fillna(0)
df['Total Minor Injuries'] = pd.to_numeric(df['Total Minor Injuries'], errors='coerce').fillna(0)
df['Total Uninjured'] = pd.to_numeric(df['Total Uninjured'], errors='coerce').fillna(0)
```

Now I check for duplicate rows. If I find any, I will remove them.

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

The date column is text, like 12/31/2007. I will change it to a real date. A few rows have no date, so I will drop these rows. They are only 7 rows.

```python
df['Event Date'] = pd.to_datetime(df['Event Date'], errors='coerce')
df = df.dropna(subset=['Event Date'])
```

I will make a Year column and a Month column from the date. They will help me in the reports later.

```python
df['Year'] = df['Event Date'].dt.year
df['Month'] = df['Event Date'].dt.month
```

The Make column is not clean. The same company is written in different ways, like CESSNA and Cessna. I will make them all the same way.

```python
df['Make'] = df['Make'].str.title()
```

I will make one new column with all the injuries together.

```python
df['Total Injuries'] = df['Total Fatal Injuries'] + df['Total Serious Injuries'] + df['Total Minor Injuries']
```

Now I will make small tables for the star schema in SQL. First I make the aircraft table. Every aircraft gets an ID number.

```python
aircraft = df[['Make', 'Model']].drop_duplicates().reset_index(drop=True)
aircraft['Aircraft ID'] = aircraft.index + 1
aircraft.head()
```

I will put the aircraft ID in the big table too. The big table is the fact table.

```python
df = df.merge(aircraft, on=['Make', 'Model'], how='left')
```

I will do the same for the location. Every location gets an ID number, and I put it in the big table too.

```python
location = df[['Location']].drop_duplicates().reset_index(drop=True)
location['Location ID'] = location.index + 1
df = df.merge(location, on='Location', how='left')
location.head()
```

Now I save the clean data in one excel file. Every table is in a different sheet. I will upload this file to SQL later.

```python
with pd.ExcelWriter('clean accidents.xlsx') as writer:
    df.to_excel(writer, sheet_name='Accidents', index=False)
    aircraft.to_excel(writer, sheet_name='Aircraft', index=False)
    location.to_excel(writer, sheet_name='Location', index=False)
```
